<h2><center> Mengidentifikasi Informasi Dataset </center></h2>

```python
train = pd.read_csv("train.csv")
train.head()
```

<h2><center>Cleaning Data</center><h2>

```python
# Menampilkan ringkasan informasi dari dataset
train.info()
```

```python
# Menampilkan statistik deskriptif dari dataset
train.describe(include="all")
```

<h2><center>Mengatasi Missing Value</center><h2>

```python
# Memeriksa jumlah nilai yang hilang di setiap kolom
missing_values = train.isnull().sum()
missing_values[missing_values > 0]
```

```python
less = missing_values[missing_values < 1000].index
over = missing_values[missing_values >= 1000].index
```

```python
# Contoh mengisi nilai yang hilang dengan median untuk kolom numerik
numeric_features = train[less].select_dtypes(include=['number']).columns
train[numeric_features] = train[numeric_features].fillna(train[numeric_features].median())
```

```python
 # Contoh mengisi nilai yang hilang dengan mode untuk kolom kategori
kategorical_features = train[less].select_dtypes(include=['object']).columns

for column in kategorical_features:
    train[column] = train[column].fillna(train[column].mode()[0])
```

```python
# Menghapus kolom dengan terlalu banyak nilai yang hilang
df = train.drop(columns=over)
```

```python
missing_values = df.isnull().sum()
missing_values[missing_values > 0]
```

<h2><center>Mengatasi Outliers</center></h2>

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

for feature in numeric_features:
    plt.figure(figsize=(10, 6))
    sns.boxplot(x=df[feature])
    plt.title(f'Box Plot of {feature}')
    plt.show()
```

```python
# Contoh sederhana untuk mengidentifikasi outliers menggunakan IQR
Q1 = df[numeric_features].quantile(0.25)
Q3 = df[numeric_features].quantile(0.75)
IQR = Q3 - Q1
```

```python
# Filter dataframe untuk hanya menyimpan baris yang tidak mengandung outliers pada kolom numerik
condition = ((df[numeric_features] < (Q1 - 1.5 * IQR)) | (df[numeric_features] > (Q3 + 1.5 * IQR))).any(axis=1)
df_filtered_numeric = df.loc[condition, numeric_features]

# Menggabungkan kembali dengan kolom kategorikal
categorical_features = df.select_dtypes(include=['object']).columns
df = pd.concat([df_filtered_numeric, df.loc[condition, categorical_features]], axis=1)
```

Dengan menggunakan LabelEncoder, jumlah kolom yang ada pada data tidak mengalami perubahan sama sekali. Hal ini karena LabelEncoder hanya mengubah value dari masing-masing fitur yang ia proses.

```python
import sklearn
```

```python
# Memisahkan fitur X dan target Y
X = df_lencoder.drop(columns=['SalePrice'])
y = df_lencoder['SalePrice']
```

```python
from sklearn.model_selection import train_test_split

# Membagi dataset menjadi training dan testing
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
# Menghitung panjang/jumlah data
print("Jumlah data: ", len(X))

# Menghitung panjang/jumlah data pada x_train
print("Jumlah data latih: ", len(x_train))

# Menghitung panjang/jumlah pada data x_test
print("Jumlah data test: ", len(x_test))
```

```python
#Melatih model 1 dengan algoritma Least Angle Regression
from sklearn import linear_model
lars = linear_model.Lars(n_nonzero_coefs=1).fit(x_train, y_train)

# Melatih model 2 dengan algoritma Linear Regression
from sklearn.linear_model import LinearRegression
LR = LinearRegression().fit(x_train, y_train)

# Melatih model 3 dengan algoritma Gradient Boosting Regressor
from sklearn.ensemble import GradientBoostingRegressor
GBR = GradientBoostingRegressor(random_state=184)
GBR.fit(x_train, y_train)
```

### Evaluasi

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Evaluasi pada model LARS
pred_lars = lars.predict(x_test)
mae_lars = mean_absolute_error(y_test, pred_lars)
mse_lars = mean_squared_error(y_test, pred_lars)
r2_lars = r2_score(y_test, pred_lars)

# Membuat dictionary untuk menyimpan hasil evaluasi
data = {
    'MAE': [mae_lars],
    'MSE': [mse_lars],
    'R2': [r2_lars]
}
# Konversi dictionary menjadi DataFrame
df_results = pd.DataFrame(data, index=['Lars'])
df_results
```

```python
# Evaluasi pada model Linear Regression
pred_LR = LR.predict(x_test)
mae_LR = mean_absolute_error(y_test, pred_LR)
mse_LR = mean_squared_error(y_test, pred_LR)
r2_LR = r2_score(y_test, pred_LR)

# Menambahkan hasil evaluasi LR ke DataFrame
df_results.loc['Linear Regression'] = [mae_LR, mse_LR, r2_LR]
df_results
```

```python
# Evaluasi pada model Gradient Boosting Regressor
pred_GBR = GBR.predict(x_test)
mae_GBR = mean_absolute_error(y_test, pred_GBR)
mse_GBR = mean_squared_error(y_test, pred_GBR)
r2_GBR = r2_score(y_test, pred_GBR)

# Menambahkan hasil evaluasi GBR ke DataFrame
df_results.loc['GradientBoostingRegressor'] = [mae_GBR, mse_GBR, r2_GBR]
df_results
```

```python
import joblib
 
# Menyimpan model ke dalam file
joblib.dump(GBR, 'gbr_model.joblib')
```

### EVALUASI HASIL


Menurut Mathotaarachchi et al. (2024) dalam jurnal Advanced Machine Learning Techniques for Predictive Modeling of Property Prices, MAE (Mean Absolute Error) digunakan untuk mengukur rata-rata selisih absolut antara nilai aktual dan nilai prediksi, sedangkan MSE (Mean Squared Error) menghitung rata-rata kuadrat selisih tersebut. Pada hasil pengujian notebook, model Gradient Boosting Regressor memperoleh MAE sebesar 0,217991 dan MSE sebesar 0,125475, lebih rendah dibandingkan Linear Regression dengan MAE 0,269327 dan MSE 0,410253, serta LARS dengan MAE 0,729010 dan MSE 0,893037. Hal ini menunjukkan bahwa Gradient Boosting Regressor menghasilkan kesalahan prediksi paling rendah di antara ketiga model yang diuji. Penjelasan teori MAE dan MSE terdapat pada bagian 3.7 Performance Evaluation.

Sementara itu, menurut Matien et al. (2026) dalam jurnal Prediksi Harga Rumah Menggunakan Linear Regression pada Perumahan Kecamatan Paguyangan, R² (Coefficient of Determination) digunakan untuk mengukur seberapa besar variasi nilai aktual yang dapat dijelaskan oleh model regresi. Hasil notebook menunjukkan bahwa Gradient Boosting Regressor memperoleh R² sebesar 0,865678, lebih tinggi dibandingkan Linear Regression sebesar 0,560821 dan LARS sebesar 0,043997. Nilai tersebut menunjukkan bahwa Gradient Boosting Regressor mampu menjelaskan sekitar 86,57% variasi target pada data pengujian, sedangkan sisanya belum dapat dijelaskan oleh model. Dengan demikian, berdasarkan ketiga metrik evaluasi, Gradient Boosting Regressor menjadi model terbaik di antara ketiga model yang diuji. Perlu diperhatikan bahwa nilai MAE dan MSE tersebut mengikuti skala target yang digunakan dalam notebook, sehingga tidak dapat langsung diartikan sebagai kesalahan harga dalam rupiah apabila target telah dinormalisasi atau distandardisasi.

Letak kutipan: Bagian 3.7 – Performance Evaluation. Pada bagian ini, penulis menjelaskan metrik evaluasi regresi, yaitu MSE, MAE, RMSE, dan R², beserta rumusnya. Teori MAE dan MSE digunakan sebagai dasar untuk menjelaskan perbandingan kesalahan prediksi ketiga model dalam notebook
