```python

# 1. Import Library
import pandas as pd
import matplotlib.pyplot as plt

# 2. Informasi Dataset
nama_dataset = "Seeds Dataset"
url_dataset = "https://archive.ics.uci.edu/dataset/236/seeds"

print("Nama Dataset:", nama_dataset)
print("Sumber Dataset:", url_dataset)

# 3. Membaca Dataset
# Dataset asli menggunakan pemisah tab (\t)

kolom = [
    "area",
    "perimeter",
    "compactness",
    "length",
    "width",
    "asymmetry",
    "length_groove",
    "class"
]

df = pd.read_csv(
    "seeds_dataset.txt",
    sep=r"\s+",
    header=None,
    names=kolom,
    engine="python"
)

# 4. Informasi Dataset
print("\nJumlah Baris:", df.shape[0])
print("Jumlah Kolom:", df.shape[1])

print("\nInformasi Dataset:")
df.info()

# 5. Menampilkan Data
print("\nLima Data Pertama:")
display(df.head())

# 6. Identifikasi Kolom
print("\nNama Kolom:")
print(df.columns.tolist())

# 7. Identifikasi Target
target = "class"

print("\nKolom Target:", target)
print("Jumlah Kategori:", df[target].nunique())

print("\nDistribusi Kategori:")
print(df[target].value_counts().sort_index())

# 8. Identifikasi Jenis Klasifikasi
jumlah_kategori = df[target].nunique()

if jumlah_kategori == 2:
    jenis_klasifikasi = "Klasifikasi Biner"
elif jumlah_kategori > 2:
    jenis_klasifikasi = "Klasifikasi Multikelas"
else:
    jenis_klasifikasi = "Tidak dapat ditentukan"

print("\nJenis Klasifikasi:", jenis_klasifikasi)

# 9. Visualisasi Distribusi Kategori
df[target].value_counts().sort_index().plot(
    kind="bar",
    title="Distribusi Kategori Dataset Seeds",
    xlabel="Kategori Varietas Gandum",
    ylabel="Jumlah Data"
)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# 10. Narasi Identifikasi Dataset
print("""
NARASI IDENTIFIKASI DATASET

Dataset Seeds merupakan dataset yang berasal dari UCI Machine
Learning Repository. Dataset ini berisi pengukuran karakteristik
biji gandum, seperti area, perimeter, compactness, panjang kernel,
lebar kernel, asimetri, dan panjang alur kernel.

Dataset ini memiliki 210 data dan 8 kolom, yang terdiri dari
7 fitur dan 1 kolom target.

Dataset Seeds termasuk klasifikasi multikelas karena memiliki
tiga kategori target, yaitu 1, 2, dan 3 yang mewakili tiga
varietas gandum: Kama, Rosa, dan Canadian.

Setiap data hanya memiliki satu kategori varietas gandum.
Oleh karena itu, dataset ini termasuk klasifikasi multikelas,
bukan klasifikasi multilabel.

Jumlah data pada dataset ini memenuhi ketentuan tugas karena
memiliki lebih dari 150 baris.
""")
```
