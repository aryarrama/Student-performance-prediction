```python

# IDENTIFIKASI DATASET KLASIFIKASI MULTILABEL
# Dataset: Amphibians

import pandas as pd

# ============================================================
# 1. MEMBACA DATASET
# ============================================================

# Lokasi dataset
dataset_path = "amphibians/dataset.csv"

# URL dataset UCI
url_dataset = "https://archive.ics.uci.edu/dataset/528/amphibians"

# Membaca dataset dengan pemisah titik koma
raw = pd.read_csv(
    dataset_path,
    sep=";",
    header=None
)

# Baris pertama berisi tipe data
feature_types = raw.iloc[0].tolist()

# Baris kedua berisi nama kolom
kolom = raw.iloc[1].tolist()

# Mengambil data mulai dari baris ketiga
df = raw.iloc[2:].copy()
df.columns = kolom

# Konversi kolom berdasarkan tipe datanya
for col, tipe in zip(kolom, feature_types):
    if tipe in ["Integer", "Numerical", "Ordinal"] or tipe.startswith("Label"):
        df[col] = pd.to_numeric(df[col], errors="coerce")


# ============================================================
# 2. IDENTIFIKASI DATASET
# ============================================================

print("Nama Dataset: Amphibians")
print("URL Dataset:", url_dataset)
print("Jumlah Baris:", df.shape[0])
print("Jumlah Kolom:", df.shape[1])

print("\nInformasi Dataset:")
df.info()

print("\nLima Data Pertama:")
display(df.head())


# ============================================================
# 3. MENENTUKAN KOLOM LABEL
# ============================================================

label_columns = [
    "Green frogs",
    "Brown frogs",
    "Common toad",
    "Fire-bellied toad",
    "Tree frog",
    "Common newt",
    "Great crested newt"
]

print("\nKolom Label:")
print(label_columns)

print("\nJumlah Label:", len(label_columns))


# ============================================================
# 4. DISTRIBUSI LABEL
# ============================================================

print("\nDistribusi Label:")
print(df[label_columns].sum())


# ============================================================
# 5. JUMLAH LABEL PADA SETIAP DATA
# ============================================================

jumlah_label_per_data = df[label_columns].sum(axis=1)

print("\nJumlah Label pada Setiap Data:")
print(jumlah_label_per_data.value_counts().sort_index())


# ============================================================
# 6. CONTOH DATA DENGAN LEBIH DARI SATU LABEL
# ============================================================

print("\nContoh Data yang Memiliki Lebih dari Satu Label:")

display(
    df.loc[
        jumlah_label_per_data > 1,
        ["ID"] + label_columns
    ].head(10)
)


# ============================================================
# 7. NARASI IDENTIFIKASI
# ============================================================

print("""
Narasi:

Dataset Amphibians termasuk klasifikasi multilabel karena satu data
dapat memiliki lebih dari satu label secara bersamaan. Dataset ini
berisi karakteristik lingkungan dan kondisi lokasi pengamatan yang
digunakan untuk memprediksi keberadaan beberapa jenis amphibians.

Pada identifikasi ini, terdapat 7 kolom label, yaitu Green frogs,
Brown frogs, Common toad, Fire-bellied toad, Tree frog, Common newt,
dan Great crested newt.

Nilai 1 menunjukkan bahwa jenis amphibian tersebut terdapat pada
suatu data, sedangkan nilai 0 menunjukkan bahwa jenis tersebut tidak
terdapat pada data tersebut.

Satu lokasi dapat memiliki beberapa jenis amphibian sekaligus.
Sebagai contoh, satu data dapat memiliki label Brown frogs = 1,
Common toad = 1, dan Tree frog = 1 secara bersamaan.

Oleh karena itu, dataset Amphibians termasuk klasifikasi multilabel,
karena satu data dapat menghasilkan lebih dari satu label.
""")


# ============================================================
# 8. KESIMPULAN
# ============================================================

is_multilabel = (df[label_columns].sum(axis=1) > 1).any()

print("\nKesimpulan:")

if is_multilabel:
    print("Dataset Amphibians termasuk klasifikasi MULTILABEL.")
    print(
        "Alasannya: terdapat data yang memiliki lebih dari satu "
        "label bernilai 1 secara bersamaan."
    )
else:
    print("Dataset tidak menunjukkan karakteristik multilabel.")
```
