```python

import pandas as pd
import numpy as np

# 1. Loading Dataset
url_dataset = "https://www.kaggle.com/datasets/shivamb/netflix-shows"
file_path = "netflix_titles.csv"

df = pd.read_csv(file_path)

print("Nama Dataset: Netflix Movies and TV Shows")
print("URL Dataset:", url_dataset)
print("Jumlah Baris:", df.shape[0])
print("Jumlah Kolom:", df.shape[1])

# 2. Informasi Dataset
print("\nInformasi Dataset:")
df.info()

print("\nLima Data Pertama:")
display(df.head())

# Pisahkan string genre menjadi list
genres = df['listed_in'].dropna().str.split(', ')

# Dapatkan daftar semua genre unik
unique_genres = sorted(list(set([genre for sublist in genres for genre in sublist])))

# Buat kolom biner untuk masing-masing genre (One-Hot Encoding untuk Multi-Label)
for genre in unique_genres:
    df[f"label_{genre}"] = df['listed_in'].apply(lambda x: 1 if isinstance(x, str) and genre in x.split(', ') else 0)

label_columns = [col for col in df.columns if col.startswith("label_")]

print("\nKolom Label Multi-Label (Genre):")
print(label_columns)

print("\nJumlah Total Label Genre Unik:", len(label_columns))

print("\nDistribusi 10 Label Genre Teratas:")
print(df[label_columns].sum().sort_values(ascending=False).head(10))
```

Narasi:
Dataset Netflix Movies and TV Shows tergolong ke dalam permasalahan klasifikasi multi-label.
Dalam dataset ini, setiap judul film atau acara TV dapat dikategorikan ke dalam lebih dari satu
genre sekaligus pada kolom 'listed_in'.

Sebagai contoh, satu judul film dapat memiliki kombinasi label seperti:
- label_Dramas = 1
- label_International Movies = 1
- label_Comedies = 1

Karena satu baris data (satu tayangan film/TV show) dapat memiliki beberapa label kategori/genre
secara bersamaan, maka dataset ini memenuhi kriteria klasifikasi multi-label.
