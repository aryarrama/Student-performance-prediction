```python
# IDENTIFIKASI DATASET KLASIFIKASI MULTILABEL
# Dataset: Flags

import pandas as pd

url_dataset = "https://archive.ics.uci.edu/dataset/40/flags"

kolom = [
    "name",
    "landmass",
    "zone",
    "area",
    "population",
    "language",
    "religion",
    "bars",
    "stripes",
    "colors",
    "red",
    "green",
    "blue",
    "gold",
    "white",
    "black",
    "orange",
    "mainhue",
    "circles",
    "crosses",
    "saltires",
    "quarters",
    "sunstars",
    "crescent",
    "triangle",
    "icon",
    "animate",
    "text",
    "topleft",
    "botright"
]

df = pd.read_csv(
    "flags/flag.data",
    header=None,
    names=kolom
)

print("Nama Dataset: Flags")
print("URL Dataset:", url_dataset)
print("Jumlah Baris:", df.shape[0])
print("Jumlah Kolom:", df.shape[1])

print("\nInformasi Dataset:")
df.info()

print("\nLima Data Pertama:")
display(df.head())

label_columns = [
    "red",
    "green",
    "blue",
    "gold",
    "white",
    "black",
    "orange"
]

print("\nKolom Label:")
print(label_columns)

print("\nJumlah Label:", len(label_columns))

print("\nDistribusi Label:")
print(df[label_columns].sum())
```

Narasi:
Dataset Flags termasuk klasifikasi multilabel karena satu data
dapat memiliki lebih dari satu label secara bersamaan. Dataset ini
berisi karakteristik bendera dari berbagai negara.

Pada identifikasi ini, warna merah, hijau, biru, emas, putih,
hitam, dan orange digunakan sebagai label. Satu bendera dapat
memiliki beberapa warna sekaligus. Sebagai contoh, sebuah bendera
dapat memiliki label red = 1, white = 1, dan blue = 1 secara
bersamaan.

Oleh karena itu, dataset Flags termasuk klasifikasi multilabel,
karena satu data dapat menghasilkan lebih dari satu label.
