```python
# IDENTIFIKASI DATASET KLASIFIKASI BINER
# Dataset: Mushroom

import pandas as pd

url_dataset = "https://archive.ics.uci.edu/dataset/73/mushroom"

kolom = [
    "class", "cap-shape", "cap-surface", "cap-color",
    "bruises", "odor", "gill-attachment", "gill-spacing",
    "gill-size", "gill-color", "stalk-shape", "stalk-root",
    "stalk-surface-above-ring", "stalk-surface-below-ring",
    "stalk-color-above-ring", "stalk-color-below-ring",
    "veil-type", "veil-color", "ring-number", "ring-type",
    "spore-print-color", "population", "habitat"
]

df = pd.read_csv(
    "agaricus-lepiota.data",
    header=None,
    names=kolom
)

print("Nama Dataset: Mushroom")
print("URL Dataset:", url_dataset)
print("Jumlah Baris:", df.shape[0])
print("Jumlah Kolom:", df.shape[1])

print("\nInformasi Dataset:")
df.info()

print("\nLima Data Pertama:")
display(df.head())

target = "class"

print("\nKolom Target:", target)
print("Jumlah Kategori:", df[target].nunique())
print("Kategori Target:")
print(df[target].value_counts())

print("""
Narasi:
Dataset Mushroom termasuk klasifikasi biner karena targetnya
memiliki dua kategori, yaitu edible (dapat dimakan) dan poisonous
(beracun). Data ini digunakan untuk mengidentifikasi kategori jamur
berdasarkan karakteristik fisiknya.
""")
```
