```python

# Identifikasi Dataset Breast Cancer Wisconsin
# Jenis: Binary Classification

import pandas as pd
from sklearn.datasets import load_breast_cancer

# Sumber dataset
url = "https://archive.ics.uci.edu/dataset/17/breast+cancer+wisconsin+diagnostic"

# Memuat dataset
data = load_breast_cancer(as_frame=True)
df = data.frame.copy()

# Mengubah nama target
df = df.rename(columns={"target": "diagnosis"})
df["diagnosis"] = df["diagnosis"].map({
    0: "malignant",
    1: "benign"
})

# Identifikasi dataset
print("Nama Dataset: Breast Cancer Wisconsin")
print("URL Sumber:", url)
print("Jumlah Baris:", df.shape[0])
print("Jumlah Kolom:", df.shape[1])

print("\nNama Kolom:")
print(df.columns.tolist())

print("\nKategori Label:")
print(df["diagnosis"].unique())

print("\nJumlah Kategori:")
print(df["diagnosis"].nunique())

print("\nDistribusi Label:")
print(df["diagnosis"].value_counts())

display(df.head())
```
