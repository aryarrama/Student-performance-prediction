```python
# IDENTIFIKASI DATASET KLASIFIKASI BINER
# Dataset: Indian Liver Patient

import pandas as pd

url_dataset = "https://archive.ics.uci.edu/dataset/225/ilpd%2Bindian%2Bpatient%2Bdataset"

kolom = [
    "Age", "Gender", "Total_Bilirubin", "Direct_Bilirubin",
    "Alkaline_Phosphotase", "Alamine_Aminotransferase",
    "Aspartate_Aminotransferase", "Total_Proteins",
    "Albumin", "Albumin_and_Globulin_Ratio", "Selector"
]

df = pd.read_csv(
    "Indian Liver Patient Dataset (ILPD).csv",
    header=None,
    names=kolom
)

print("Nama Dataset: Indian Liver Patient")
print("URL Dataset:", url_dataset)
print("Jumlah Baris:", df.shape[0])
print("Jumlah Kolom:", df.shape[1])

print("\nInformasi Dataset:")
df.info()

print("\nLima Data Pertama:")
display(df.head())

target = "Selector"

print("\nKolom Target:", target)
print("Jumlah Kategori:", df[target].nunique())
print("Kategori Target:")
print(df[target].value_counts())

print("""
Narasi:
Dataset Indian Liver Patient termasuk klasifikasi biner karena
memiliki dua kategori target, yaitu pasien dengan penyakit hati
dan pasien tanpa penyakit hati. Dataset ini menggunakan beberapa
indikator kesehatan untuk mengidentifikasi kategori pasien.
""")
```
