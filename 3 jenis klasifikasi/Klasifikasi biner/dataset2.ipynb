```python

# Identifikasi Dataset Bank Marketing
# Jenis: Binary Classification

import pandas as pd
import requests
import zipfile
import io

# URL sumber dataset
url = "https://archive.ics.uci.edu/static/public/222/bank+marketing.zip"

# Download dataset
response = requests.get(url)
response.raise_for_status()

# Membaca file ZIP
with zipfile.ZipFile(io.BytesIO(response.content)) as z:
    file_name = next(
        name for name in z.namelist()
        if name.endswith("bank-full.csv")
    )
    df = pd.read_csv(z.open(file_name), sep=";")

# Identifikasi dataset
print("Nama Dataset: Bank Marketing")
print("URL Sumber:", url)
print("Jumlah Baris:", df.shape[0])
print("Jumlah Kolom:", df.shape[1])

print("\nNama Kolom:")
print(df.columns.tolist())

print("\nKategori Label:")
print(df["y"].unique())

print("\nJumlah Kategori:")
print(df["y"].nunique())

print("\nDistribusi Label:")
print(df["y"].value_counts())

display(df.head())
```
